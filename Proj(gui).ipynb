```python
import tkinter as tk
from tkinter import messagebox,ttk
import pandas as pd
import joblib
```

```python
model = joblib.load('classifier.joblib')
label_encoder_platform = joblib.load('label_encoder_platform.joblib')
label_encoder_gender = joblib.load('label_encoder_gender.joblib')
label_encoder_emotion = joblib.load('label_encoder_emotion.joblib')
```

```python
class EmotionPredictorApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Emotion Predictor")
        self.root.geometry("450x400")
        self.root.configure(bg="#f0f0f0")
        self.age_var = tk.StringVar()
        self.gender_var = tk.StringVar()
        self.platform_var = tk.StringVar()
        self.comments_received_var = tk.StringVar()
        self.likes_received_var = tk.StringVar()
        self.messages_sent_var = tk.StringVar()
        self.posts_per_day_var = tk.StringVar()
        self.daily_usage_time_var = tk.StringVar()
        self.create_widgets()

    def create_widgets(self):
        font_label = ("Helvetica", 12)
        font_entry = ("Helvetica", 10)

        tk.Label(self.root, text="Age", font=font_label, bg="#f0f0f0").grid(row=0, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.age_var, font=font_entry).grid(row=0, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Gender", font=font_label, bg="#f0f0f0").grid(row=1, column=0, padx=10, pady=5, sticky=tk.W)
        gender_frame = tk.Frame(self.root, bg="#f0f0f0")
        gender_frame.grid(row=1, column=1, padx=10, pady=5)
        tk.Radiobutton(gender_frame, text="Male", variable=self.gender_var, value="Male", bg="#f0f0f0").pack(side=tk.LEFT)
        tk.Radiobutton(gender_frame, text="Female", variable=self.gender_var, value="Female", bg="#f0f0f0").pack(side=tk.LEFT)

        tk.Label(self.root, text="Platform", font=font_label, bg="#f0f0f0").grid(row=2, column=0, padx=10, pady=5, sticky=tk.W)
        platform_dropdown = ttk.Combobox(self.root, textvariable=self.platform_var, font=font_entry)
        platform_dropdown['values'] = ("Instagram", "Twitter", "Facebook", "LinkedIn", "Snapchat", "Telegram")
        platform_dropdown.grid(row=2, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Comments Received Per Day", font=font_label, bg="#f0f0f0").grid(row=3, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.comments_received_var, font=font_entry).grid(row=3, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Likes Received Per Day", font=font_label, bg="#f0f0f0").grid(row=4, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.likes_received_var, font=font_entry).grid(row=4, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Messages Sent Per Day", font=font_label, bg="#f0f0f0").grid(row=5, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.messages_sent_var, font=font_entry).grid(row=5, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Posts Per Day", font=font_label, bg="#f0f0f0").grid(row=6, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.posts_per_day_var, font=font_entry).grid(row=6, column=1, padx=10, pady=5)

        tk.Label(self.root, text="Daily Usage Time (minutes)", font=font_label, bg="#f0f0f0").grid(row=7, column=0, padx=10, pady=5, sticky=tk.W)
        tk.Entry(self.root, textvariable=self.daily_usage_time_var, font=font_entry).grid(row=7, column=1, padx=10, pady=5)

        predict_button = ttk.Button(self.root, text="Predict", command=self.predict_emotion)
        predict_button.grid(row=8, column=0, columnspan=2, pady=20)

    def predict_emotion(self):
        try:
            input_data = {
                "Age": int(self.age_var.get()),
                "Platform": self.platform_var.get(),
                "Comments_Received_Per_Day": int(self.comments_received_var.get()),
                "Likes_Received_Per_Day": int(self.likes_received_var.get()),
                "Gender": self.gender_var.get(),
                "Messages_Sent_Per_Day": int(self.messages_sent_var.get()),
                "Posts_Per_Day": int(self.posts_per_day_var.get()),
                "Daily_Usage_Time (minutes)": int(self.daily_usage_time_var.get())
            }
            input_df = pd.DataFrame([input_data])
            input_df = input_df[[
                "Age",
                "Platform",
                "Comments_Received_Per_Day",
                "Likes_Received_Per_Day",
                "Gender",
                "Messages_Sent_Per_Day",
                "Posts_Per_Day",
                "Daily_Usage_Time (minutes)"
            ]]
            input_df["Platform"] = label_encoder_platform.transform(input_df["Platform"])
            input_df["Gender"] = label_encoder_gender.transform(input_df["Gender"])
            prediction = model.predict(input_df)[0]
            result = label_encoder_emotion.inverse_transform([prediction])[0]
            messagebox.showinfo("Prediction Result", f"Dominant Emotion: {result}")
        except Exception as e:
            messagebox.showerror("Error", str(e))

if __name__ == "__main__":
    root = tk.Tk()
    app = EmotionPredictorApp(root)
    root.mainloop()
```
